```python
df["Installs"] = (
    df["Installs"].astype(str).str.replace(",","",regex=False).str.replace("+","",regex=False).str.extract(r"(\d+)").astype(float).fillna(0).astype(int)
)
df["Reviews"] = pd.to_numeric(df["Reviews"],errors="coerce").fillna(0).astype(int)
df[["Category","Installs","Reviews"]].head()
```

```python
if df["Rating"].isnull().any():
    df["Rating"] = df["Rating"].fillna(df["Rating"].median())
df.isnull().sum()
```

```python
#aggregating by installs
category_installs = (df.groupby("Category", as_index=False)["Installs"].sum().sort_values("Installs", ascending=False).reset_index(drop=True)
                    )
category_installs.head(10)
```

```python
excluded_alphabets = tuple("ACGS")
filtered = category_installs[~category_installs["Category"].str.upper().str.startswith(excluded_alphabets)].reset_index(drop=True)
top_5 = filtered.head(5).copy()
print("Top 5 categories after exclusion:\n")
print(top_5)
```

```python
top_5_list = top_5["Category"].tolist()
top_5_list
```

```python
df_top_5 = df[df["Category"].isin(top_5_list)].copy()
print(df_top_5.shape)
df_top_5.head()
```

```python
#creating a choropleth
countries = ["IND","USA","GBR","AUS","CAN","ARE","DEU"]
weights = np.array([0.22,0.25,0.12,0.10,0.11,0.10],dtype=float)
weights = weights/weights.sum()
rows=[]
for cat in top_5_list:
    total_cat_installs = df_top_5.loc[df_top_5["Category"]==cat,"Installs"].sum()
    split = (total_cat_installs*weights).astype(int)
    remainder=int(total_cat_installs)-int(split.sum())
    if remainder>0:
        split[weights.argmax()]+=remainder
    for iso, val in zip(countries,split):
        rows.append(
            {"category":cat,"country_iso3":iso,"installs":int(val)}
        )
installs_by_country = pd.DataFrame(rows)
print(installs_by_country.head(10))
print(installs_by_country.groupby("category")["installs"].sum().reset_index())
```

```python
installs_by_country.to_csv("installs_by_country.csv", index=False)
print("File saved")
```

```python
import plotly.express as px
import ipywidgets as widgets
from IPython.display import display

def plot_category(category):
    df_cat = installs_by_country[installs_by_country["category"] == category]
    fig = px.choropleth(df_cat, locations = "country_iso3",color="installs",color_continuous_scale="Viridis",projection="natural earth",title=f"Global Installs-{category}",)
    fig.show()

dropdown = widgets.Dropdown(options = top_5_list, description = "Category", value=top_5_list[0])
widgets.interact(plot_category, category=dropdown);
```

```python
import plotly.express as px
import plotly.graph_objects as go
import ipywidgets as widgets
from IPython.display import display

def plot_category_with_highlight(category, threshold=1_000_000, show_highlight=True):

    df_cat = installs_by_country[installs_by_country["category"] == category].copy()
    if df_cat.empty:
        print(f"No rows found for category: {category}")
        return

    fig = px.choropleth(
        df_cat,
        locations="country_iso3",
        color="installs",
        color_continuous_scale="Viridis",
        projection="natural earth",
        title=f"Global Installs — {category} (Highlight > {int(threshold):,})",
    )

    if show_highlight:
        thr = int(threshold)
        df_hi = df_cat[df_cat["installs"] > thr]
        if not df_hi.empty:
            fig.add_trace(
                go.Choropleth(
                    locations=df_hi["country_iso3"],
                    z=df_hi["installs"],
                    colorscale=[[0, "rgba(0,0,0,0)"], [1, "rgba(0,0,0,0)"]],
                    showscale=False,
                    marker_line_width=2.5,
                    marker_line_color="red",
                    hoverinfo="skip",
                )
            )

    fig.update_layout(margin=dict(l=0, r=0, t=50, b=0), height=600)
    fig.show()

categories = sorted(installs_by_country["category"].dropna().unique().tolist())
w_category = widgets.Dropdown(options=categories, value=categories[0], description="Category:")
w_threshold = widgets.IntText(value=1_000_000, description="Threshold:")
w_toggle = widgets.ToggleButton(value=True, description="Show Highlight")

ui = widgets.HBox([w_category, w_threshold, w_toggle])
out = widgets.interactive_output(
    plot_category_with_highlight,
    {"category": w_category, "threshold": w_threshold, "show_highlight": w_toggle},
)

display(ui, out)
```
